FUNCTION PROGRAMS

Addition & difference

```python
def calc(a=0,b=0):
    return a+b, a-b

print(calc(10,3))
```

Multiply using keyword arguments

```python
def multiply(a,b,c):
    return a*b*c

print(multiply(a=12,b=23,c=34))
```

*find_maximum(args)

```python
def find_maximum(*args):
    return max(args)

print(find_maximum(5,8,2,9))
```

**summarize(kwargs)

```python
def summarize(**kwargs):
    for k,v in kwargs.items():
        print(k,":",v)

summarize(name="John", age=25)
```

**Add numbers using *args and kwargs

```python
def add_all(*args, **kwargs):
    total = 0
    for v in args:
        total += v
    for v in kwargs.values():
        total += v
    return total

print(add_all(1,2,a=3,b=4))
```
