LAMBDA PROGRAMS

```python
sum10 = (lambda n: sum(range(1,n+1)))(10)
print(sum10)
```

```python
neg = lambda x: x < 0
print(neg(-3))
```

```python
pos = lambda x: x > 0
print(pos(5))
```

```python
length = lambda s: len(s)
print(length("hello"))
```

```python
add1 = list(map(lambda x: x+1, [2,4,6,8]))
print(add1)
```

```python
less5 = lambda x: x < 5
print(less5(3))
```

```python
odd = lambda x: x % 2 != 0
print(odd(7))
```

```python
sub5 = lambda x: x - 5
print(sub5(10))
```

```python
gt10 = lambda x: x > 10
print(gt10(20))
```

```python
div3 = lambda x: x % 3 == 0
print(div3(9))
```

```python
sum2 = lambda a,b: a+b
print(sum2(5,6))
```

```python
less15 = lambda x: x < 15
print(less15(10))
```
