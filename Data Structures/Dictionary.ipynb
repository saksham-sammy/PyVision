DICTIONARY PROGRAMS

1. Create dictionary


```python
d = {"name": "Alice", "age": 21}
print(d)
```


2. Add key-value


```python
d = {"name": "Alice"}
d["city"] = "Delhi"
print(d)
```


3. Remove using del & pop


```python
d = {"name":"Alice", "age":20}
del d["age"]
print(d)

d["city"] = "Mumbai"
print(d.pop("city"))
print(d)
```


4. Access value


```python
d = {"name":"Alice", "age":20}
print(d["name"])
```

5. Check key exists


```python
d = {"name":"Alice"}
print("name" in d)
```

6. Get keys, values, items

```python
d = {"name":"Alice", "age":22}
print(d.keys())
print(d.values())
print(d.items())
```

7. Update value

```python
d = {"age":20}
d["age"] = 25
print(d)
```
