SET PROGRAMS

1. Create set

```python
s = {1,2,3,4}
print(s)
```

2. Add, remove, discard


```python
s = {1,2,3}
s.add(4)
s.remove(2)
s.discard(10)
print(s)
```

3. Union, intersection, difference


```python
a = {1,2,3}
b = {3,4,5}
print(a | b)
print(a & b)
print(a - b)
print(a ^ b)
```

4. Subset, superset

```python
a = {1,2}
b = {1,2,3,4}
print(a.issubset(b))
print(b.issuperset(a))
```

5. Check membership


```python
s = {1,2,3}
print(2 in s)
```
