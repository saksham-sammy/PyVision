LOOPS & CONDITIONS


1. Print 1–10

```python
for i in range(1,11):
    print(i)
```

2. Check positive/negative/zero

```python
n = int(input())
if n > 0:
    print("Positive")
elif n < 0:
    print("Negative")
else:
    print("Zero")
```

3. Even or odd

```python
n = int(input())
print("Even" if n%2==0 else "Odd")
```

4. Odd numbers 1–20

```python
for i in range(1,21):
    if i % 2 != 0:
        print(i)
```

5. Pattern 1

```python
for i in range(1,5):
    print("*" * i)
```

6. Pattern 2



```python
for i in range(1,6):
    for j in range(1,i+1):
        print(j, end="")
    print()
```

7. 10 to 1

```python
for i in range(10,0,-1):
    print(i)
```
