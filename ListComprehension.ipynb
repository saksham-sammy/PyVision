LIST COMPREHENSION PROGRAMS

1. First 20 even numbers

```python
evens = [2*i for i in range(1,21)]
print(evens)
```

2. Squares of 1–5

```python
squares = [x*x for x in range(1,6)]
print(squares)
```

3. First 5 odd numbers

```python
odds = [2*i-1 for i in range(1,6)]
print(odds)
```

4. Multiply 1–5 by 3

```python
res = [x*3 for x in range(1,6)]
print(res)
```

5. Double each element

```python
lst = [1,2,3]
print([x*2 for x in lst])
```

6. Cubes 1–10

```python
print([x**3 for x in range(1,11)])
```

7. Squares of even numbers

```python
print([x*x for x in range(1,21) if x%2==0])
```
